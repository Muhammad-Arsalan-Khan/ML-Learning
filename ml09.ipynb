Outliers

```python
import seaborn as sns
import numpy as np
import pandas as pd

data = pd.DataFrame({ 'age': [23, 45, 56, 67, 34, 23, 45, 56, 67, 34, 150, 200] })
data
sns.histplot(data, kde=True)
# sns.scatterplot(x=range(len(data)), y=data)
data['mean'] = np.mean(data['age'])
data['std'] = np.std(data['age'])
data['z-score'] = np.abs((data['age'] - data['mean']) / data['std']) 
data = data[data['z-score'] < 1.5]  # Remove outliers with z-score > 3
data
```
